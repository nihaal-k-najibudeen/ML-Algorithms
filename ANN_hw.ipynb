```python
df.drop(["StudentID","Age","Gender"],axis=1,inplace=True)
df
```

```python
x = df.drop("GradeClass",axis=1)
y = df["GradeClass"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
import tensorflow as tf
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
ann = Sequential()
ann.add(Dense(11,activation='relu'))
ann.add(Dense(10,activation='relu'))
ann.add(Dense(5,activation='relu'))
ann.add(Dense(5,activation='softmax'))
```

```python
ann.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(x_train,y_train,epochs=100,batch_size=20)
```

```python
ann.evaluate(x_test,y_test)
```

```python
y_pred = ann.predict(scaler.transform([[0,3,10.5,2,0,4,1,0,0,0,4.5]]))
output = np.argmax(y_pred)
output
```
